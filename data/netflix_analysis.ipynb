```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("netflix_titles.csv")

movies = df[df['type'] == 'Movie']

plt.hist(movies['duration'].str.replace(' min', '').astype(int), bins=20)
plt.title("Distribution of Movie Duration")
plt.xlabel("Minutes")
plt.ylabel("Count")
plt.show()

movies['main_genre'] = movies['listed_in'].str.split(',').str[0]
movies['main_genre'].value_counts().head(10).plot(kind='bar')
plt.title("Top 10 Movie Genres")
plt.show()

movies['country'].value_counts().head(10).plot(kind='bar')
plt.title("Top Countries Producing Netflix Movies")
plt.show()

movies['release_year'].value_counts().sort_index().plot()
plt.title("Trend of Movies Released Over the Years")
plt.xlabel("Year")
plt.ylabel("Movie Count")
plt.show()

print("Total Movies:", len(movies))
```
